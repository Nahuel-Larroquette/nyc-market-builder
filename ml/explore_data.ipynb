## Data Cleaning

```python
df = df.drop(columns=['Unnamed: 0'], errors="ignore")
df.head()
```

```python
df['SQFT_ESTIMATED'] = df['PROPERTYSQFT'] == 2184.207862
df['SQFT_ESTIMATED'].value_counts()
```

```python
median_sqft = df.loc[~df['SQFT_ESTIMATED'], 'PROPERTYSQFT'].median()
df.loc[df['SQFT_ESTIMATED'], 'PROPERTYSQFT'] = median_sqft

df['PROPERTYSQFT'].describe()
```

```python
import numpy as np
df['LOG_PRICE'] = np.log(df['PRICE'])
df['LOG_PRICE'].hist(bins=50)
```
